## Distance, Speed, and Time

The relationship between distance, speed, and time is represented by:

d = v × t

Where:
- d = distance
- v = speed
- t = time

Depending on which value is unknown, the equation can be rearranged:

- Distance: d = v × t
- Speed: v = d ÷ t
- Time: t = d ÷ v

```python
from calculations import (
    calculate_distance,
    calculate_speed,
    calculate_time,
    calculate_scaled_distance
)

# Example: How long does sunlight take to reach Earth?

distance = 149_600_000
speed = 299_792

time = calculate_time(distance, speed)

print("Sunlight Travel Time")
print("--------------------")
print(f"Time: {time:.2f} seconds")
print(f"Time: {time / 60:.2f} minutes")
```
