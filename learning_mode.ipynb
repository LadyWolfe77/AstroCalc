## Distance, Speed, and Time

The relationship between distance, speed, and time is represented by:

d = v × t

Where:
- d = distance
- v = speed
- t = time

Depending on which value is unknown, the equation can be rearranged:

- Distance: d = v × t
- Speed: v = d ÷ t
- Time: t = d ÷ v

```python
from calculations import (
    calculate_distance,
    calculate_speed,
    calculate_time,
    calculate_scaled_distance
)

# Example: How long does sunlight take to reach Earth?

distance = 149_600_000
speed = 299_792

time = calculate_time(distance, speed)

print("Sunlight Travel Time")
print("--------------------")
print(f"Time: {time:.2f} seconds")
print(f"Time: {time / 60:.2f} minutes")
```

Astronomy Lab Notes: Scaling the Solar System
1. What are we trying to calculate?

The actual Solar System is much too large to draw at its real size. Instead, we're shrinking all the distances by the same proportion so the planets can fit on a strip of paper.

This is called a scale model.

The important idea is that every object must use the same scale. If Pluto is at the end of the paper, all the other planets must be placed at the correct proportional distances from the Sun.

2. Equation 1: Calculate the scale factor
Scale Factor= Paper Length (cm) / Pluto’s Actual Distance (km)

	​
What does this tell us?

The scale factor tells us how many centimeters on our paper represent one kilometer in the actual Solar System.

For this lab, Pluto's average distance from the Sun is:
5.91×10**9 km

Let's say your paper strip measures 150 cm.

Step 1: Write the known values.

Paper length = 150 cm

Pluto's actual distance = 5.91×10**9 km

Step 2: Substitute the values.

Scale Factor= 150 / 5.91×10**9

	​


Step 3: Calculate.

Scale Factor≈2.54×10**−8

Result
2.54×10**−8 cm/km

The units are important: centimeters per kilometer.

3. Equation 2: Calculate each planet's scaled distance

Once we know the scale factor, we can calculate where each planet belongs on the paper.

Scaled Distance=Actual Distance×Scale Factor

Why do we multiply?

The scale factor tells us how much the actual distance needs to shrink. Multiplying the planet's actual distance by that factor converts it into the smaller paper distance.

Let's use Earth as our example.

Earth's average distance from the Sun is:

1.50×10**8 km

Step 1: Write the known values.

Earth's actual distance = 1.50×10**8 km

Scale factor = 2.54×10**−8 cm/km

Step 2: Substitute into the equation.

Scaled Distance=(1.50×10**8)(2.54×10**−8)

Step 3: Multiply.

Scaled Distance≈3.81 cm
Result

Earth should be marked approximately 3.81 cm from the Sun on our 150 cm paper strip.

Notice that the kilometers cancel out, leaving centimeters. That's how we know the resulting unit is correct.

4. How the two equations work together

Think of the process as two steps:

First: Find the scale factor

Paper Length / Actual Pluto Distance
	​
Second: Find each planet's paper position

Actual Planet Distance×Scale Factor

Final result: The planet's distance from the Sun on your paper, in centimeters.

These are example results using a 150 cm strip. Your actual answers will depend on your measured paper length.

5. A few things worth remembering

The Sun is at 0 cm because it is our starting point.

Pluto is at the end of the strip because its distance establishes the scale.

The inner planets are much closer together than the outer planets.

The asteroid belt is a range, so we calculate its inner and outer edges separately.

These calculations use average orbital distances (semi-major axes), not the planets' exact positions on a particular day.

```python
from calculations import calculate_scaled_distance

# Enter the measured length of your paper strip
strip_length = float(input("Enter strip length in centimeters: "))

# Actual distances from the Sun in kilometers
objects = {
    "Sun": 0,
    "Mercury": 5.83e7,
    "Venus": 1.08e8,
    "Earth": 1.50e8,
    "Mars": 2.27e8,
    "Asteroid Belt (inner edge)": 3.28e8,
    "Asteroid Belt (outer edge)": 4.77e8,
    "Jupiter": 7.78e8,
    "Saturn": 1.43e9,
    "Uranus": 2.87e9,
    "Neptune": 4.50e9,
    "Pluto": 5.91e9
}

# Calculate and display scaled positions
print("\nSOLAR SYSTEM SCALED DISTANCES")
print("--------------------------------")

for name, distance in objects.items():
    scaled = calculate_scaled_distance(distance, strip_length)
    print(f"{name:30} {scaled:8.2f} cm")
```
